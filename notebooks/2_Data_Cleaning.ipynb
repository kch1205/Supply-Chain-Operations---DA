```python
gross_sales = df_clean["Sales"].sum()
total_discounts = df_clean["Order Item Discount"].sum()
net_sales = df_clean["Order Item Total"].sum()

print("Gross sales:", round(gross_sales, 2))
print("Total discounts:", round(total_discounts, 2))
print("Net sales:", round(net_sales, 2))
print(
    "Sales reconciliation difference:",
    round(net_sales - (gross_sales - total_discounts), 2)
)

profit_match = (
    df_clean["Benefit per order"]
    == df_clean["Order Profit Per Order"]
)

print("Matching profit records:", profit_match.sum())
print("Mismatching profit records:", (~profit_match).sum())
```

```python
df_clean["Calculated_Net_Sales"] = (
    df_clean["Sales"] - df_clean["Order Item Discount"]
)

df_clean["Sales_Difference"] = (
    df_clean["Order Item Total"]
    - df_clean["Calculated_Net_Sales"]
)

mismatch_rows = df_clean[
    df_clean["Sales_Difference"].abs() > 0.01
]

print("Rows with sales differences:", len(mismatch_rows))
print(
    "Total absolute difference:",
    round(mismatch_rows["Sales_Difference"].abs().sum(), 2)
)

display(
    mismatch_rows[
        [
            "Sales",
            "Order Item Discount",
            "Order Item Total",
            "Sales_Difference"
        ]
    ].head(10)
)
```

```python
diff = df_clean["Sales_Difference"].round(2)

print("Rows with any difference:", (diff != 0).sum())
print("Rows differing by exactly 0.01:", (diff.abs() == 0.01).sum())
print("Rows differing by more than 0.01:", (diff.abs() > 0.01).sum())

print(
    "Net total difference:",
    round(df_clean["Sales_Difference"].sum(), 2)
)
```
