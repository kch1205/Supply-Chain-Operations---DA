```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Load the cleaned dataset
data_path = Path("../data/processed/apl_logistics_cleaned.csv")
df = pd.read_csv(data_path)

print("Revenue & Profit Analysis")
print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
gross_sales = df["Sales"].sum()
net_sales = df["Order Item Total"].sum()
total_discounts = df["Order Item Discount"].sum()
total_profit = df["Order Profit Per Order"].sum()

profit_margin = total_profit / net_sales * 100

print("Gross Sales:", round(gross_sales, 2))
print("Net Sales:", round(net_sales, 2))
print("Total Discounts:", round(total_discounts, 2))
print("Total Profit:", round(total_profit, 2))
print("Profit Margin (%):", round(profit_margin, 2))
```

```python
gross_sales = df["Sales"].sum()
net_sales = df["Order Item Total"].sum()
total_discounts = df["Order Item Discount"].sum()
total_profit = df["Order Profit Per Order"].sum()

profit_margin = (total_profit / net_sales) * 100

print("REVENUE & PROFIT SUMMARY")
print("-" * 35)
print("Gross Sales:", round(gross_sales, 2))
print("Net Sales:", round(net_sales, 2))
print("Total Discounts:", round(total_discounts, 2))
print("Total Profit:", round(total_profit, 2))
print("Profit Margin (%):", round(profit_margin, 2))
```

```python
import matplotlib.pyplot as plt

labels = ["Gross Sales", "Net Sales", "Total Profit"]
values = [gross_sales, net_sales, total_profit]

plt.figure(figsize=(9, 5))
bars = plt.bar(labels, values)

plt.title("APL Logistics: Revenue and Profit Overview")
plt.ylabel("Amount")
plt.xlabel("Financial Metrics")
plt.ticklabel_format(axis="y", style="plain")

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:,.0f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
market_summary = df.groupby("Market").agg(
    Total_Net_Sales=("Order Item Total", "sum"),
    Total_Profit=("Order Profit Per Order", "sum"),
    Total_Discounts=("Order Item Discount", "sum"),
    Total_Orders=("Order Customer Id", "count")
).reset_index()

market_summary = market_summary.sort_values(
    by="Total_Profit", ascending=False
)

display(market_summary.round(2))
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    market_summary["Market"],
    market_summary["Total_Profit"]
)

plt.title("Total Profit by Market")
plt.xlabel("Market")
plt.ylabel("Total Profit")
plt.xticks(rotation=30)
plt.ticklabel_format(axis="y", style="plain")
plt.tight_layout()
plt.show()
```

```python
market_summary["Profit_Margin_Percent"] = (
    market_summary["Total_Profit"]
    / market_summary["Total_Net_Sales"]
) * 100

market_summary = market_summary.sort_values(
    by="Profit_Margin_Percent",
    ascending=False
)

display(
    market_summary[
        [
            "Market",
            "Total_Net_Sales",
            "Total_Profit",
            "Profit_Margin_Percent"
        ]
    ].round(2)
)
```

```python
margin_data = market_summary.sort_values(
    by="Profit_Margin_Percent",
    ascending=False
)

plt.figure(figsize=(9, 5))

bars = plt.bar(
    margin_data["Market"],
    margin_data["Profit_Margin_Percent"]
)

plt.title("Profit Margin by Market")
plt.xlabel("Market")
plt.ylabel("Profit Margin (%)")

for bar in bars:
    height = bar.get_height()
    plt.text(
    bar.get_x() + bar.get_width() / 2,
    height,
    f"{height:.2f}%",
    ha="center",
    va="bottom"
)

plt.tight_layout()
plt.show()
```
